```python
# BMI Calculator
# Calculates Body Mass Index (BMI)
# and displays the weight category.


# Making a reusable block of code -calculate_bmi(function)
# weight and height are called parameters
def calculate_bmi(weight, height):
    return weight / (height ** 2)           # BMI formula


def bmi_category(bmi):                                                                                      
    if bmi <= 18.49:                        # 18.5 is considered normal weight
        return "Underweight"
    elif bmi <= 24.99:                      # 25 is considered overwirght  /  elif bmi < 25: 64x1.6
        return "Normal Weight"
    elif bmi <= 29.99:                      # 30 is considered obese
        return "Overweight"
    else:
        return "Obese"                      # => 30 obese

def main():
    print("=" * 40)                         # Props
    print("BMI CALCULATOR")
    print("=" * 40)                         # Props

    try:                                    # keyword used for exception handling/ attempt the code
        weight = float(input("Enter your weight (kg): "))
        print(f"Weight: {weight:.2f} kg")   # Display weight
         
        height = float(input("Enter your height (m): "))
        print(f"Height: {height:.2f} m")                    # Display height

        if weight <= 0 or height <= 0:
            print("Weight and height must be greater than zero.")
            return

        bmi = calculate_bmi(weight, height)

        print(f"\nBMI: {bmi:.2f}")
        print(f"Category: {bmi_category(bmi)}")

    except ValueError:                      # handle the error if something goes wrong
        print("Invalid input. Please enter numbers only.")
        
main()

#The American Cancer Society 
#located in Atlanta, Georgia
# Underweight: BMI is less than 18.5 
# Normal weight: BMI is 18.5 to 24.9 
# Overweight: BMI is 25 to 29.9 
# Obesity: BMI is 30 or more
```
